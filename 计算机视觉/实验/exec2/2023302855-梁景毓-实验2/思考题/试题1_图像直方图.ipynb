# 试题1：请用OpenCV和matplotlib取出一张训练集中的图像，并绘制图像直方图。（初级）

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import Image, ImageSequence
import os
```

## 定义辅助函数

```python
def load_multipage_tiff(path):
    """加载多页TIFF图像"""
    return np.array([np.array(p) for p in ImageSequence.Iterator(Image.open(path))])

def plot_histogram(image, title="Histogram"):
    """
    绘制图像直方图
    
    参数:
        image: 输入图像（灰度图）
        title: 图表标题
    """
    plt.figure(figsize=(10, 6))
    
    # 绘制直方图
    plt.hist(image.ravel(), bins=256, range=[0, 256], color='gray', alpha=0.7)
    plt.title(title, fontsize=14, fontweight='bold')
    plt.xlabel('Pixel Value', fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    plt.grid(True, alpha=0.3)
    
    # 添加统计信息
    mean_val = np.mean(image)
    std_val = np.std(image)
    min_val = np.min(image)
    max_val = np.max(image)
    
    stats_text = f'Mean: {mean_val:.2f}\nStd: {std_val:.2f}\nMin: {min_val}\nMax: {max_val}'
    plt.text(0.02, 0.98, stats_text, transform=plt.gca().transAxes,
             fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    
    plt.tight_layout()
    return plt
```

## 主程序

```python
# 数据集路径
data_dir = './data'
train_volume_path = os.path.join(data_dir, 'train-volume.tif')

# 加载训练集图像
print("正在加载训练集图像...")
images = load_multipage_tiff(train_volume_path)
print(f"训练集图像形状: {images.shape}")

# 选择第一张图像进行分析
image = images[0]
print(f"第一张图像形状: {image.shape}")
print(f"图像数据类型: {image.dtype}")
```

## 绘制原图和直方图

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# 显示原图
axes[0].imshow(image, cmap='gray')
axes[0].set_title('Original Image', fontsize=14, fontweight='bold')
axes[0].axis('off')

# 绘制直方图
axes[1].hist(image.ravel(), bins=256, range=[0, 256], color='steelblue', alpha=0.7, edgecolor='black')
axes[1].set_title('Image Histogram', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Pixel Value', fontsize=12)
axes[1].set_ylabel('Frequency', fontsize=12)
axes[1].grid(True, alpha=0.3)

# 添加统计信息
mean_val = np.mean(image)
std_val = np.std(image)
min_val = np.min(image)
max_val = np.max(image)

stats_text = f'Mean: {mean_val:.2f}\nStd: {std_val:.2f}\nMin: {min_val}\nMax: {max_val}'
axes[1].text(0.02, 0.98, stats_text, transform=axes[1].transAxes,
             fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig('./思考题/histogram_result.png', dpi=300, bbox_inches='tight')
print("直方图已保存到: ./思考题/histogram_result.png")
plt.show()
```

## 打印统计信息

```python
print("\n图像统计信息:")
print(f"  均值: {mean_val:.2f}")
print(f"  标准差: {std_val:.2f}")
print(f"  最小值: {min_val}")
print(f"  最大值: {max_val}")
print(f"  中位数: {np.median(image):.2f}")

# 分析直方图特征
print("\n直方图特征分析:")
print("  - 该数据集的图像直方图比较连续")
print("  - 没有出现明显的波谷")
print("  - 说明前景和背景的像素值分布重叠较多")
print("  - 使用传统的基于统计的阈值分割算法可能效果不佳")
```
